# Q1 Single Feature Noise

- We will first copy the original dataset for each of the percentage noises (20% & 40%) so that we dont acidentally make changes to the orignial dataset

- For the Single Feature Noise will be adding **NaN Values** to the first feature radius which will be **Radius1** column in the WDBC dataset.

```python
"""
 Create copies of the original features to avoid 
 modifying the original dataset for each percentage
 p = [20% , 40%]
"""
single_feature_20 = X.copy()
single_feature_40 = X.copy()

# Set a random seed for results stay consistent
np.random.seed(42)

# 20% Noise Level
# Calculate the number of samples to change to NaN in the radius1 column
num_samples_20 = int(0.20 * len(X)) # 20% of 569 which is 113

# Randomly select 20% indices from the dataframe to be turned into NaNs
indices_20 = np.random.choice(X.index, num_samples_20, replace=False)

# Using .loc to replace the 'radius1' values
# radius1 is the first feature in the dataset
single_feature_20.loc[indices_20, 'radius1'] = np.nan

# 40% Noise Level
# Calculate the number of samples to change to NaN in the radius1 column
num_samples_40 = int(0.40 * len(X)) # 40% of 569 which is 227

# Randomly select 40% indices from the dataframe to be turned into NaNs
indices_40 = np.random.choice(X.index, num_samples_40, replace=False)

# Replace 'radius1' values with NaN for the 40% noise
single_feature_40.loc[indices_40, 'radius1'] = np.nan

# Print the count of missing values
print(f"Total NaNs with 20% of the data points, replaced on the first radius feature (radius1) column: {single_feature_20['radius1'].isnull().sum()}")
print(f"Total NaNs with 40% of the data points, replaced on the first radius feature (radius1) column: {single_feature_40['radius1'].isnull().sum()}")
```

```python
# Checking if the Radius1 column got NaN (Null values)
print("Showing the 20% NaN values in the first radius feature (Radius1):")
single_feature_20.isnull().sum()
```

```python
# Checking if the Radius1 column got NaN (Null values)
print("Showing the 40% NaN values in the first radius feature (Radius1):")
single_feature_40.isnull().sum()
```

# Q1 Random Feature Noise
- We will first copy the original dataset for each of the percentage noises (20% & 40%) so that we dont acidentally make changes to the orignial dataset

- For the Random Feature Noise variant it will select a feature uniformly at random and replace its value with ‘nan’

```python
"""
 Create copies of the original features to avoid 
 modifying the original dataset for each percentage
 p = [20% , 40%]
"""
random_feature_20 = X.copy()
random_feature_40 = X.copy()

# 20% Random Feature Noise
# Randomly select 20% of noise to the dataset setting to NaNs
rows_to_noise_20 = np.random.choice(X.index, int(0.20 * len(X)), replace=False)

# Loop through each selected row and pick a random column to change to NaNs
for row in rows_to_noise_20:
    # Pick a random column name from the list of all features
    random_col = np.random.choice(X.columns)
    # Set that specific cell to NaN
    random_feature_20.at[row, random_col] = np.nan

# 40% Random Feature Noise
# Randomly select 40% of noise to the dataset setting to NaNs
rows_to_noise_40 = np.random.choice(X.index, int(0.40 * len(X)), replace=False)

# Loop through each selected row for the 40% variant
for row in rows_to_noise_40:
    # Pick a random column name
    random_col = np.random.choice(X.columns)
    # Set that specific cell to NaN
    random_feature_40.at[row, random_col] = np.nan

# Verification: Check total NaNs across the entire dataframe
print(f"Total NaNs with 20% of the data points, replaced on the random features: {random_feature_20.isnull().sum().sum()}")
print(f"Total NaNs with 40% of the data points, replaced on the random features: {random_feature_40.isnull().sum().sum()}")
```

```python
# Checking if random columns got NaN (Null values)
print(random_feature_20.isnull().sum())
print("Total NaN for all the columns: ",random_feature_20.isnull().sum().sum())
```

```python
# Checking if random columns got NaN (Null values)
print(random_feature_40.isnull().sum())
print("Total NaN for all the columns: ",random_feature_40.isnull().sum().sum())
```

# Q1 Imputations and Performance

 - We are going to use the ‘impute’ function of Scikit-learn: https://scikit-learn.org/1.5/modules/impute.html
 
 - Using **SimpleImputer vs IterativeImputer** 
 
 - Apply imputation to train and test sets independently.

 - Use Stratified 5-Fold Cross-Validation.

 - Calculate the F1 Score (average +/- standard deviation).

 - Compare 3 cases: Original noisy dataset, SimpleImputer, and IterativeImputer. 
 
 ## Implementation Process for both Single Feature & Random Feature (20%, 40%) :
 - First I would use the new datasets with 'NaN' values
   - Single Feature: **single_feature_20 & single_feature_40** 
   - Random Feature: **random_feature_20 & random_feature_40**
   
   
 - **Handling NaNs:** LogisticRegression cannot handle NaN values, the "Noisy Original" was established by filling missing values with 0 using .fillna(0).
- **Imputation Strategy:** We compared **SimpleImputer** which uses the mean of the column vs **IterativeImputer** which treats the missing feature as a target and predicts it using the other 29 features.
- **Validation:** Ensure results were robust, we utilized Stratified 5-Fold Cross-Validations
- **Performance**  Measure the average +/- standard deviation of the F1 score


I add **Standard Scaling** from Scikit learn to standardizes data features to have a mean of 0 and a standard deviation of 1, without it when I ran max_iter= 100 for the logistic regression it didn't converge and got a warning.
ref: https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html

# Q1 Single Feature Noise Imputations for 20% Noise

```python
# Showing the Dataset with the NaNs to the first Radius 'Radius1' feature (20%)
single_feature_20
```

```python
from sklearn.preprocessing import StandardScaler

# Create a arrays to store F1 scores for each fold
single_20_results = {
    'singleNoisy_20': [],
    'singleSimpleImputer_20': [],
    'singleIterativeImputer_20': []
}

# Initialize Stratified 5-Fold CV
cross_val = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scaler = StandardScaler()

print("Single Feature Noise 20% 5-Fold Results: ")

# Tracking each fold for each case for outputing it
for fold_idx, (train_idx, test_idx) in enumerate(cross_val.split(single_feature_20, y), 1):
    # Splitting the training data and testing data
    X_train, X_test = single_feature_20.iloc[train_idx], single_feature_20.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    
    # convert into 1D
    y_train_flat = y_train.values.flatten()

    # Noisy Original
    # Logistic Regression can't work with NaN so I use .fillna(0)
    X_train_zero = X_train.fillna(0)
    X_test_zero = X_test.fillna(0)
    
    # Scaling the train and test data
    X_train_scaled = scaler.fit_transform(X_train_zero)
    X_test_scaled = scaler.transform(X_test_zero)
    
    # Default max_iter=100
    clf_noisy = LogisticRegression(max_iter=100) 
    clf_noisy.fit(X_train_scaled, y_train_flat)
    
    # storing the f1 score that it predicts "Malignant"
    f1_noisy = f1_score(y_test, clf_noisy.predict(X_test_scaled), pos_label='M')
    # Add results to the singleNoisy_20 array
    single_20_results['singleNoisy_20'].append(f1_noisy)

    # SimpleImputer
    simple_imp = SimpleImputer() # default parameters
    # fit transform train and test seperately
    X_train_simple = simple_imp.fit_transform(X_train)
    X_test_simple = simple_imp.transform(X_test)
    
    # Scaling after imputation
    X_train_simple_scaled = scaler.fit_transform(X_train_simple)
    X_test_simple_scaled = scaler.transform(X_test_simple)
    
    # Default max_iter=100 
    clf_simple = LogisticRegression(max_iter=100)
    clf_simple.fit(X_train_simple_scaled, y_train_flat)
    
    # storing the f1 score that it predicts "Malignant"
    f1_simple = f1_score(y_test, clf_simple.predict(X_test_simple_scaled), pos_label='M')
    # Add results to the singleSimpleImputer_20 array
    single_20_results['singleSimpleImputer_20'].append(f1_simple)

    # IterativeImputer
    # used random state = 42 for consistent and same output each run
    iter_imp = IterativeImputer(random_state=42)
    X_train_iter = iter_imp.fit_transform(X_train)
    X_test_iter = iter_imp.transform(X_test)
    
    # Scaling after imputation
    X_train_iter_scaled = scaler.fit_transform(X_train_iter)
    X_test_iter_scaled = scaler.transform(X_test_iter)
    
    # Default max_iter=100 
    clf_iter = LogisticRegression(max_iter=100)
    clf_iter.fit(X_train_iter_scaled, y_train_flat)
    
    # storing the f1 score that it predicts "Malignant"
    f1_iter = f1_score(y_test, clf_iter.predict(X_test_iter_scaled), pos_label='M')
    # Add results to the singleIterativeImputer_20 array
    single_20_results['singleIterativeImputer_20'].append(f1_iter)
    
    # Output the F1 scores for the current fold
    print(f"Fold {fold_idx}: Noisy={f1_noisy:.4f}, Simple={f1_simple:.4f}, Iterative={f1_iter:.4f}")

print("\nPerformance Metrics (Average +/- standard deviation):")

# Calculate and printing the average of f1 scores and standard deviation
for method, scores in single_20_results.items():
    mean_f1 = np.mean(scores)
    std_f1 = np.std(scores)
    print(f"{method}: Average F1 Score = {mean_f1:.4f} & Standard Deviation = +/- {std_f1:.4f}")
```

```python
from sklearn.preprocessing import StandardScaler

# Create a arrays to store F1 scores for each fold
random_40_results = {
    'randomNoisy_40': [],
    'randomSimpleImputer_40': [],
    'randomIterativeImputer_40': []
}

# Initialize Stratified 5-Fold CV
cross_val = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scaler = StandardScaler()

print("Random Feature Noise 40% 5-Fold Results: ")

# Tracking each fold for each case for outputing it
for fold_idx, (train_idx, test_idx) in enumerate(cross_val.split(random_feature_40, y), 1):
    # Splitting the training data and testing data
    X_train, X_test = random_feature_40.iloc[train_idx], random_feature_40.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    # convert into 1D
    y_train_flat = y_train.values.flatten()

    # Noisy Original
    # Logistic Regression can't work with NaN so I use .fillna(0)
    X_train_zero = X_train.fillna(0)
    X_test_zero = X_test.fillna(0)
    
    # Scaling the filled data
    X_train_scaled = scaler.fit_transform(X_train_zero)
    X_test_scaled = scaler.transform(X_test_zero)
    
    # Default max_iter=100
    clf_noisy = LogisticRegression(max_iter=100) 
    clf_noisy.fit(X_train_scaled, y_train_flat)
    # storing the f1 score that it predicts
    f1_noisy = f1_score(y_test, clf_noisy.predict(X_test_scaled), pos_label='M')
    # Add results to the randomNoisy_40 array
    random_40_results['randomNoisy_40'].append(f1_noisy)

    # SimpleImputer
    simple_imp = SimpleImputer() # default parameters
    # fit transform train and test seperately
    X_train_simple = simple_imp.fit_transform(X_train)
    X_test_simple = simple_imp.transform(X_test)
    
    # Scaling after imputation
    X_train_simple_scaled = scaler.fit_transform(X_train_simple)
    X_test_simple_scaled = scaler.transform(X_test_simple)
    
    # Default max_iter=100 
    clf_simple = LogisticRegression(max_iter=100)
    # convert into 1D
    clf_simple.fit(X_train_simple_scaled, y_train_flat)
    # storing the f1 score that it predicts
    f1_simple = f1_score(y_test, clf_simple.predict(X_test_simple_scaled), pos_label='M')
    # Add results to the randomSimpleImputer_40 array
    random_40_results['randomSimpleImputer_40'].append(f1_simple)

    # IterativeImputer
    # used random state = 42 for consistent and same output each run
    iter_imp = IterativeImputer(random_state=42)
    X_train_iter = iter_imp.fit_transform(X_train)
    X_test_iter = iter_imp.transform(X_test)
    
    # Scaling after imputation
    X_train_iter_scaled = scaler.fit_transform(X_train_iter)
    X_test_iter_scaled = scaler.transform(X_test_iter)
    
    # Default max_iter=100 
    clf_iter = LogisticRegression(max_iter=100)
    # convert into 1D
    clf_iter.fit(X_train_iter_scaled, y_train_flat)
    # storing the f1 score that it predicts
    f1_iter = f1_score(y_test, clf_iter.predict(X_test_iter_scaled), pos_label='M')
    # Add results to the randomIterativeImputer_40 array
    random_40_results['randomIterativeImputer_40'].append(f1_iter)
    
    # Output the F1 scores for the current fold
    print(f"Fold {fold_idx}: Noisy={f1_noisy:.4f}, Simple={f1_simple:.4f}, Iterative={f1_iter:.4f}")

print("\nPerformance Metrics (Average +/- standard deviation):")

# Calculate and printing the average of f1 scores and standard deviation
for method, scores in random_40_results.items():
    mean_f1 = np.mean(scores)
    std_f1 = np.std(scores)
    print(f"{method}: Average F1 Score = {mean_f1:.4f} & Standard Deviation = +/- {std_f1:.4f}")
```

# Q2 Dimensionality reduction 
We are going to compare how two different dimensionality reduction techniques affect the performance.
## Scope
- **Singular Value Decomposition:** The Singular Value Decomposition (SVD) is an extremely useful tool from linear algebra that can help us approximate a data matrix (such as the instance-by-feature matrix that represents our data) into a smaller dimension, hence reduce the dimensionality of the data. 

- **Autoencoder:** In certain ways, the autoencoder is the neural network alternative to the SVD, where an encoder network produces an intermediate (latent) representation of that data, and a decoder network transforms the data back to the original space, while ensuring during training that the decoded output is as close to the input as possible. 

- **plot:** for k = [2, 5, 10] being the dimensionality of the latent representation, create a figure that shows the performance as a function of k for the two dimensionality reduction techniques and draws as a straight horizontal line the average performance without dimensionality reduction. You may skip adding error-bars on that horizontal line only. 

# Getting the Original Average F1 score without dimension reduction

- First I wanted to see the orignial f1 score before the dimensional reduction to be able draws as a straight horizontal line the average performance without dimensionality reduction
- I used 5-fold validation and get the average of the mean and standard deviation of the original dataset

```python
# Calculate and getting Original baseline performance with the 30 features
original_f1 = []
# initializing the 5-folds
cross_val = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for train_idx, test_idx in cross_val.split(X, y):
    # Standard Split
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    # Convert to 1D
    y_train_flat = y.iloc[train_idx].values.flatten()
    
    # Scaling is crucial before Logistic Regression and SVD
    scaler = StandardScaler()
    X_tr_scaled = scaler.fit_transform(X_train)
    X_te_scaled = scaler.transform(X_test)
    
    # Fit classifier on full feature set
    clf = LogisticRegression(max_iter=100)
    clf.fit(X_tr_scaled, y_train_flat)
    
    # Store F1 score for the 'M' (Malignant) class
    original_f1.append(f1_score(y.iloc[test_idx], clf.predict(X_te_scaled), pos_label='M'))

# Store the means of the 
original_avg = np.mean(original_f1)
original_std = np.std(original_f1)
print(f"Original without dimensional reduction: \n")
print(f"Average F1 score: {original_avg:.4f}")
print(f"Standard Deviation F1 score: +/- {original_std:.4f}\n")
```

# Q2 Singular Value Decomposition

- **Implementation**
    - Ref NumPy: https://numpy.org/devdocs/reference/generated/numpy.linalg.svd.html
    - I am using 'numpy.linalg.svd' provided from numPy to help reduce the dimensionallity if the dataset to k = [2,5,10]. 
    - To avoid data leakage the **Singular Value Decomposition** was compute only on the train split and project in the test instances to the space in the same way that the **Latent Semantic Analysis paper**.

- **Scaling:** Standard scaling was applied within each fold prior to SVD to ensure that features with larger magnitudes did not disproportionately influence the singular values.

```python
#  Dimensionality Reduction (SVD)

# Define the k-values 
# Storing Singular Value Decomposition Means and Standard Deviation
k_values = [2, 5, 10]
svd_means = []
svd_stds = []

# Initializing SVD Implementation Loop
for k in k_values:
    fold_scores = []
    print(f"Running SVD for k={k}...")
    
    for train_idx, test_idx in cross_val.split(X, y):
        # Extract training and testing sets for this fold
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train_flat = y.iloc[train_idx].values.flatten()
        
        # Scaling the fold seperate train and test to avoid leakage 
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)
        
        # Perform Singular Value Decomposition on the scaled training data
        # Utilize the syntax as shown in the numpy document
        # U = Left singular vectors, S = Singular values, Vh = Right singular vectors
        U, S, Vh = np.linalg.svd(X_train_scaled, full_matrices=False)
        
        # Extract the top k components (the first k rows of Vh)
        # Transpose Vh[:k] to get the projection matrix (V_k) of shape (30, k)
        Vk = Vh[:k, :].T
        
        #print("The Projection Matrix:", Vk)
        
        # Project both train and test data into the k-dimensional space
        # X_latent = X_original * V_k
        X_train_k = np.dot(X_train_scaled, Vk)
        X_test_k = np.dot(X_test_scaled, Vk)
        
        # Train the Logistic Regression model after dimension is reduced
        clf = LogisticRegression(max_iter=100)
        clf.fit(X_train_k, y_train_flat)
        
        # Evaluate performance on the projected test data
        pred = clf.predict(X_test_k)
        
        # Store F1 score for the 'M' (Malignant) class
        fold_scores.append(f1_score(y.iloc[test_idx], pred, pos_label='M'))
    
    # Store the average and standard deviation
    svd_means.append(np.mean(fold_scores))
    svd_stds.append(np.std(fold_scores))
    
    # Output the mean and standard deviation
    print(f"SVD k={k} Average F1 Score: {svd_means[-1]:.4f} and Standard Deviation +/- {svd_stds[-1]:.4f}")
```

# Q2 Autoendcoder

- **Import MLPRegressor**: Since scikit didnt have an autoencoder document, MLPRegressor was suggested to use to build and train using Sci-kit learn Multi-Layer Perceptron.
   - https://scikit-learn.org/stable/modules/generated/sklearn.neural_network.MLPRegressor.html#sklearn.neural_network.MLPRegressor-
    

- **The Structure for the Neural Network:** The autoencoder would have two hidden layers, with widths equal to  
    - (a) min(d,2*k) (where d is the original dimensionality, 'd' is 30) 
    - (b) k = [2, 5, 10] respectively and accordingly for the decoder, and with ReLU activation functions.

- Similar to the SVD, the autoencoder is only be trained on the train set and only applied to the test set. The output of the encoder will be the new latent representation of the data. 

- **Optimization:** 
  - **Convergance Warning** 'Stochastic Optimizer: Maximum iterations (200) reached and the optimization hasn't converged yet'.
  - I set max_iter=700 rather than the default of 200 iterations to help on the convergence and used a fixed random_state=42 to ensure the the neural network training remained the same output.

```python
# Dimensionality Reduction (Autoencoder)
from sklearn.neural_network import MLPRegressor

# Arrays to store the Mean F1 and Standard Deviation for the Autoencoder
autoencoder_means = []
autoencoder_stds = []
d = X.shape[1]  # Original dimensionality (30 features)

print("Starting Autoencoder Dimensionality Reduction...")

for k in k_values:
    fold_scores = []
    # hidden_layer_sizes structure per instructions: (min(d, 2k), k, min(d, 2k))
    h1_width = min(d, 2 * k)
    
    print(f"Processing k={k} (Hidden layer width: {h1_width})...")
    
    for train_idx, test_idx in cross_val.split(X, y):
        # Splitting data into the 5-fold training and testing sets
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train_flat = y.iloc[train_idx].values.flatten()
        
        # Scaling for neural network convergence
        scaler = StandardScaler()
        X_tr_scale = scaler.fit_transform(X_train)
        X_te_scale = scaler.transform(X_test)
        
        # Define the Autoencoder architecture
        # Architecture: Input -> h1_width -> k (Latent) -> h1_width -> Output
        """
        Max_iter=200 have a little warning of convergance so I increased to 500 
        to help a little
        """
        autoencoder = MLPRegressor(hidden_layer_sizes=(h1_width, k, h1_width), 
                          activation='relu', 
                          random_state=42, 
                          max_iter=700)
        
        # Train the Autoencoder to reconstruct the input
        # (X_tr_s is used as both the input and the target)
        autoencoder.fit(X_tr_scale, X_tr_scale)
        
        # Extracting the Latent Representation (the middle layer output)
        # We access the learned weights (coefs_) and biases (intercepts_)
        # coefs_[0] is Input -> h1; coefs_[1] is h1 -> Latent (k)
        W0, W1 = autoencoder.coefs_[0], autoencoder.coefs_[1]
        b0, b1 = autoencoder.intercepts_[0], autoencoder.intercepts_[1]
        
        # Manual forward pass to the 'k' layer: ReLU(X * W0 + b0) * W1 + b1
        def get_latent(data):
            h1 = np.maximum(0, np.dot(data, W0) + b0) # Apply ReLU activation
            return np.dot(h1, W1) + b1               # Latent layer output
            
        X_train_latent = get_latent(X_tr_scale)
        X_test_latent = get_latent(X_te_scale)
        
        # Train the Logistic Regression classifier on the k-dimensional latent features
        clf = LogisticRegression(max_iter=100)
        clf.fit(X_train_latent, y_train_flat)
        
        # Calculate F1 score for the Malignant 'M' class
        score = f1_score(y.iloc[test_idx], clf.predict(X_test_latent), pos_label='M')
        fold_scores.append(score)
        
    # Average the scores across the 5 folds
    autoencoder_means.append(np.mean(fold_scores))
    autoencoder_stds.append(np.std(fold_scores))
    print(f"Autoencoder k={k}: Avg F1 = {autoencoder_means[-1]:.4f} +/- {autoencoder_stds[-1]:.4f}")
```

| k=5 | 0.9501 +/- 0.0207 | 0.9266 +/- 0.0428 |

```python
# Initialize the figure size for better readability
plt.figure(figsize=(10, 6))

# Plot SVD performance with error bars
# fmt='-o' creates a solid line with circle markers
plt.errorbar(k_values, svd_means, yerr=svd_stds, label='SVD', fmt='-o', 
             capsize=5, capthick=1, color='blue', linewidth=2, markersize=8)

# Plot Autoencoder performance with error bars
# fmt='-s' creates a solid line with square markers
plt.errorbar(k_values, autoencoder_means, yerr=autoencoder_stds, label='Autoencoder', fmt='-s', 
             capsize=5, capthick=1, color='green', linewidth=2, markersize=8)

# Draw a horizontal dashed line for the baseline performance (no reduction)
plt.axhline(y=original_avg, color='red', linestyle='--', linewidth=2, 
            label=f'Original without Dimensional Reduction: {original_avg:.4f}')

# Labeling and design
plt.title('Performance vs. Latent Dimension (k)', fontsize=14, fontweight='bold')
plt.xlabel('Latent Dimension (k)', fontsize=12)
plt.ylabel('Average F1 Score (Malignant Class)', fontsize=12)

# Ensure the x-axis only shows the specific k values we tested
plt.xticks(k_values)

# Adding a legend with a slight shadow for clarity
plt.legend(loc='lower right', frameon=True, shadow=True)

# Add a grid to make it easier to read specific values
plt.grid(True, linestyle='--', alpha=0.7)

# Adjust the layout to prevent clipping of labels
plt.tight_layout()

plt.show()
```

- **Autoencoder** 
  - There is decrease in the latent dimension k=2 and k=5 then an increase up to k=10 achieving a f1 score of 0.9596.
