```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torchvision
import torch.nn as nn
import torch.optim as optim
from torchvision import transforms
import torch.nn.functional as F
from torch.utils.data import Subset, DataLoader
# clustering algorithms and silhouette score
from sklearn.cluster import KMeans, SpectralClustering
from sklearn.metrics import silhouette_score
from sklearn.model_selection import train_test_split
```

# MNIST Data Loading
- 28 x 28 pixel
- Totaling 784 pixels per image

```python
# Download the Dataset and Transform to Tensors
train_data = torchvision.datasets.MNIST('./data', train=True, download=True, transform=transforms.ToTensor())
test_data = torchvision.datasets.MNIST('./data', train=False, download=True, transform=transforms.ToTensor())

# Load the train and test data with Batch size of 64
train_loader = torch.utils.data.DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = torch.utils.data.DataLoader(test_data, batch_size=64)

# Confirming if loaded
print("Confirming the download for MNIST Dataset")
print(f"Training samples: {len(train_data)}")
print(f"Test samples: {len(test_data)}")
```

# Preparation of Dataset A (0-4)

- Download the full MNIST dataset using torchvision
- Loop through and keep indices that has for **digits 0-4 only** to create Dataset A (same as Part 2)
- Split Dataset A into:
  - **80%** for training the 2-Layer CNN
  - **20%** for clustering evaluation (this is what we will cluster)

```python
#  Creating a function to get specific digits
def get_digits(dataset, labels):
    indices = []
    for i in range(len(dataset)):
        # dataset.targets contains the digit (0-9) for each image
        if dataset.targets[i].item() in labels:
            indices.append(i)
    return indices

# Identify indices for the two groups
train_digits_A = get_digits(train_data, [0, 1, 2, 3, 4])
test_digits_A = get_digits(test_data, [0, 1, 2, 3, 4])

print(train_digits_A)
# print(test_digits_A)
```

# Splitting Dataset A into 80% CNN training and 20% clustering set

```python
train_idx, cluster_idx = train_test_split(train_digits_A, test_size=0.20, random_state=42)

# Create Subset objects for each split
dataset_A_train   = Subset(train_data, train_idx)    # 80% for CNN training
dataset_A_cluster = Subset(train_data, cluster_idx)  # 20% for clustering

# Print sizes to verify the split
print(f"Full Dataset A (0-4) size: {len(train_digits_A)}")
print(f"CNN Training set 80%: {len(dataset_A_train)}")
print(f"Clustering set 20%: {len(dataset_A_cluster)}")
```

# Creating the 2-Layer CNN
- Reusing the same 2-Layer CNN architecture from Part 2
- The CNN is split into two:
  - **Convolutional layers**: Learn the feature representations from the images
  - **Fully-connected layers**: Classify the learned representations of the **digits 0-4**
- After the training phase, we will extract the output of the convolutional layers as the **learned representations** for clustering

ref:
https://docs.pytorch.org/tutorials/beginner/blitz/neural_networks_tutorial.html
https://docs.pytorch.org/tutorials/beginner/blitz/cifar10_tutorial.html (CIFAR-10 Example)

**Architecture for 2-Layer CNN:**
- (Layer 1) Conv2d: 1 input channel, 6 output channels, 5x5 kernel → MaxPool 2x2
- (Layer 2) Conv2d: 6 input channels, 16 output channels, 5x5 kernel → MaxPool 2x2
- (FC1) Fully Connected: Input is 16 channels * 4 * 4 pixels = 256
- (FC2) Fully Connected: 120 inputs output the 5 classes

# Getting the Learned Representations from the 2-Layer CNN
I created a "get_learn_representation()" function to extract the output of the 
  convolutional layers **before** the fully-connected layers
- This vector is the **learned representation** of the image that the CNN 
  has learned during training on Dataset A
- This will be used as input to the clustering algorithms for the Learning Representation

```python
class MNIST2Layer(nn.Module):
    def __init__(self):
        super(MNIST2Layer, self).__init__()
        # Layer 1: 1 input channel, 6 output channels, 5x5 kernel
        # MNIST 28x28 -> Conv output 24x24 -> Pool output 12x12
        self.conv1 = nn.Conv2d(1, 6, 5)
        # 2x2 grid max pooling
        self.pool = nn.MaxPool2d(2, 2)
        
        # Layer 2: 6 input channels, 16 output channels, 5x5 kernel
        # 12x12 -> Conv output 8x8 -> Pool output 4x4
        self.conv2 = nn.Conv2d(6, 16, 5)
        
        # Fully Connected Layers: Input is 16 channels * 4 * 4 pixels = 256
        self.fc1 = nn.Linear(16 * 4 * 4, 120)
        self.fc2 = nn.Linear(120, 5) # 5 output classes (0-4 or 5-9)

    def forward(self, x):
        # Pass through Conv Layer 1
        conv1_out = self.conv1(x)
        # Apply ReLU activation to Conv Layer 1 output
        relu1_out = F.relu(conv1_out)
        # Apply MaxPool to reduce dimensions
        pool1_out = self.pool(relu1_out)
    
        # Pass through Conv Layer 2
        conv2_out = self.conv2(pool1_out)
        # Apply ReLU activation to Conv Layer 2 output
        relu2_out = F.relu(conv2_out)
        # Apply MaxPool to reduce spatial dimensions
        pool2_out = self.pool(relu2_out)
    
        # Flatten the output into a 1D vector for the fully connected layers
        flattened = torch.flatten(pool2_out, 1)
    
        # Pass through Fully Connected Layer 1 with ReLU activation
        fc1_out = F.relu(self.fc1(flattened))
    
        # Pass through Fully Connected Layer 2 to get final class scores
        output = self.fc2(fc1_out)
    
        return output
    
    def get_learn_representation(self, x):
        # Pass through Conv Layer 1
        conv1_out = self.conv1(x)
        # Apply ReLU activation to Conv Layer 1 output
        relu1_out = F.relu(conv1_out)
        # Apply MaxPool to reduce dimensions
        pool1_out = self.pool(relu1_out)
        
        # Pass through Conv Layer 2
        conv2_out = self.conv2(pool1_out)
        # Apply ReLU activation to Conv Layer 2 output
        relu2_out = F.relu(conv2_out)
        # Apply MaxPool to reduce spatial dimensions
        pool2_out = self.pool(relu2_out)
    
        # Flatten the output into a 1D vector
        representation = pool2_out.reshape(pool2_out.shape[0], -1)
        
        return representation

# Initialize the 2 layer model
model2 = MNIST2Layer()
print("2-Layer Architecture:")
print(model2)
```

# Training the 2-Layer CNN for representation of the digits
- Using the same 2-Layer CNN architecture from Part 2
- Training on the **80% training set** of Dataset A
- Optimizer:
  - SGD (Stochastic Gradient Descent) 
  - learning rate 0.01 
  - momentum 0.9 (average the gradients)
- Using **CrossEntropyLoss** as the loss function
- Training for **5 epochs** for MNIST to learn meaningful representations

```python
# Defining training loss function and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model2.parameters(), lr=0.01, momentum=0.9)

# Create DataLoader for the 80% CNN training set with batch size 64
train_loader_A = DataLoader(dataset_A_train, batch_size=64, shuffle=True)

# Train the 2-Layer CNN on Dataset A (digits 0-4)
# Setting train mode
model2.train()

# looping to do 5 epochs
for epoch in range(5):
    # Track loss and accuracy per epoch
    running_loss = 0.0
    correct = 0
    total = 0
    
    for digits, labels in train_loader_A:
        # Clear gradients from previous batch
        optimizer.zero_grad()
        
        # Forward pass
        outputs = model2(digits)
        
        # Calculating the loss
        loss = criterion(outputs, labels)
        
        # Backpropagate
        loss.backward()
        
        # Update weights
        optimizer.step()
        
        # Track accuracy
        pred = outputs.argmax(dim=1)
        total += labels.size(0)
        correct += (pred == labels).sum().item()
        running_loss += loss.item()
    
    print(f"Epoch {epoch+1}: Loss={running_loss/len(train_loader_A):.4f} | Accuracy={100*correct/total:.2f}%")

print("Training Complete!")
```

# Extracting Raw Data and Learned Representations from the 20% Cluster Set
- I extract two types of representations from the **20% clustering set**:
  - **Raw Data**: Each MNIST image flattened into a 784-length pixel vector (28x28 pixels)
  - **Learned Representations**: The output of the convolutional layers before 
    the fully-connected layers using "get_learn_representation()"
- Model is set to **evaluation mode** so no gradients are computed
- Both representations will be put into K-Means and Spectral Clustering

```python
# Create DataLoader for the 20% clustering set
# shuffle=False to keep order the same
cluster_loader = DataLoader(dataset_A_cluster, batch_size=64, shuffle=False)

# Storage lists for raw data and learned representations
raw_data = []
learned_representations = []

# Set model to evaluation mode
model2.eval()

with torch.no_grad():
    for digits, labels in cluster_loader:
        
        # Flatten each 28x28 digit into a 784-length pixel vector
        flattened = digits.reshape(digits.shape[0], -1)
        raw_data.append(flattened)
        
        # Pass digits through conv layers only to get learned representations
        representations = model2.get_learn_representation(digits)
        learned_representations.append(representations)
print("The representations")
print(representations)
print("\n")
# Concatenate all into single tensors using torch.cat
raw_data = torch.cat(raw_data, dim=0)
learned_representations = torch.cat(learned_representations, dim=0)

print("The raw data:")
print(raw_data)
print("\n")

print("The learned representations:")
print(learned_representations)
print("\n")
# Print shapes of the raw data and learned representations
print(f"Raw Data shape: {raw_data.shape}")
print(f"Learned Representations shape: {learned_representations.shape}")
```

# Clustering: K-Means and Spectral Clustering
- Running two clustering algorithms on two types of representations:
  - **K-Means** on Raw Data
  - **K-Means** on Learned Representations
    - ref: https://scikit-learn.org/stable/modules/generated/sklearn.cluster.KMeans.html
  - **Spectral Clustering** on Raw Data
  - **Spectral Clustering** on Learned Representations
    - Ref: https://scikit-learn.org/stable/modules/generated/sklearn.cluster.SpectralClustering.html
- For each clustering algorithm and representation type:
  - Run **10 times** with different random initializations
  - Compute **Silhouette Coefficient** for each run
    - Ref: https://scikit-learn.org/stable/modules/generated/sklearn.metrics.silhouette_score.html
   
  - Report **mean ± standard deviation** across 10 runs
- K values tested: **K = 1 to 5**

# Subsampling the 6120 (20% of clustering set) to 1000 sub samples for Clustering experiments
- The full 20% clustering set contains 6120 samples. Running Spectral Clustering on the full set took 5+ hours on my Macbook without finishing.
- Therefore, I subsampled down to 1000 samples to make the computation more feasible and fast. I say took around 5-6 hours running in the background  
  - **I also tried to run 1500 and 2000 sub sample but it took forever to output and my Macbook became slow**
- K-Means is also run on the same 1000 subsampled points (instead of the full 6120) to keep the comparison fair and consistent across all 4 combinations.
- A fixed random seed (seed=42) is used for the subsampling to ensure reproducibility

```python
# Convert tensor to array
raw_np = raw_data.numpy()
learned_np = learned_representations.numpy()

np.random.seed(42)
subsample_idx = np.random.choice(len(raw_np), size=1000, replace=False)
raw_np_sub = raw_np[subsample_idx]
learned_np_sub = learned_np[subsample_idx]

print(f"Original size: {len(raw_np)}")
print(f"Subsampled Raw Data shape: {raw_np_sub.shape}")
print(f"Subsampled Learned Representations shape: {learned_np_sub.shape}")
```

# K-Means Clustering on Raw Data

```python
# K-Means Clustering on Raw Data
# Running 10 times with different random seeds for K = 1 to 5
# K=1 is stored as NaN since Silhouette score requires at least 2 clusters

K_values = [1, 2, 3, 4, 5]
kmeans_raw_means = []
kmeans_raw_stds = []

for k in K_values:
    if k == 1:
        # Silhouette score is undefined for a single cluster
        kmeans_raw_means.append(np.nan)
        kmeans_raw_stds.append(np.nan)
        print("K=1: Silhouette score NaN it requires at least 2 clusters")
    else:
        scores = []
        for seed in range(10):
            # Run K-Means with default parameters and a different random seed each time
            km = KMeans(n_clusters=k, random_state=seed)
            labels = km.fit_predict(raw_np_sub)

            # Compute Silhouette Coefficient for this run
            score = silhouette_score(raw_np_sub, labels)
            scores.append(score)

        # Store mean and std across 10 runs
        kmeans_raw_means.append(np.mean(scores))
        kmeans_raw_stds.append(np.std(scores))
        print(f"K={k}: Mean={np.mean(scores):.4f} +/- Std={np.std(scores):.4f}")

print("K-Means on Raw Data complete!")
```

# K-Means on Learning Representation

```python
# K-Means Clustering on Learned Representations
# Running 10 times with different random seeds for K = 1 to 5
# K=1 is stored as NaN since Silhouette score requires at least 2 clusters

kmeans_learned_means = []
kmeans_learned_stds = []

for k in K_values:
    if k == 1:
        # Silhouette score is undefined for a single cluster
        kmeans_learned_means.append(np.nan)
        kmeans_learned_stds.append(np.nan)
        print("K=1: Silhouette score NaN it requires at least 2 clusters")
    else:
        scores = []
        for seed in range(10):
            # Run K-Means with default parameters and a different random seed each time
            km = KMeans(n_clusters=k, random_state=seed)
            labels = km.fit_predict(learned_np_sub)

            # Compute Silhouette Coefficient for this run
            score = silhouette_score(learned_np_sub, labels)
            scores.append(score)

        # Store mean and std across 10 runs
        kmeans_learned_means.append(np.mean(scores))
        kmeans_learned_stds.append(np.std(scores))
        print(f"K={k}: Mean={np.mean(scores):.4f} +/- Std={np.std(scores):.4f}")

print("K-Means on Learned Representations complete!")
```

# Spectral Clustering on Raw Data


```python
# Spectral Clustering on Raw Data
# Running 10 times with different random seeds for K = 1 to 5
# K=1 is stored as NaN since Silhouette score requires at least 2 clusters

spectral_raw_means = []
spectral_raw_stds = []

for k in K_values:
    if k == 1:
        # Silhouette score is undefined for a single cluster
        spectral_raw_means.append(np.nan)
        spectral_raw_stds.append(np.nan)
        print("K=1: Silhouette score NaN it requires at least 2 clusters")
    else:
        scores = []
        for seed in range(10):
            # Run Spectral Clustering with default parameters and a different random seed each time
            sc = SpectralClustering(n_clusters=k, random_state=seed)          
            labels = sc.fit_predict(raw_np_sub)

            # Compute Silhouette Coefficient for this run
            score = silhouette_score(raw_np_sub, labels)
            scores.append(score)
            
            # want to see the iteration if it running
            print(f"K={k}, Seed={seed+1}/10 done")


        # Store mean and std across 10 runs
        spectral_raw_means.append(np.mean(scores))
        spectral_raw_stds.append(np.std(scores))
        print(f"K={k}: Mean={np.mean(scores):.4f} +/- Std={np.std(scores):.4f}")

print("Spectral Clustering on Raw Data complete!")
```

# Spectral Clustering on Learning Representation

```python
# Spectral Clustering on Learned Representations
# Running 10 times with different random seeds for K = 1 to 5
# K=1 is stored as NaN since Silhouette score requires at least 2 clusters

K_values = [1, 2, 3, 4, 5]
spectral_learned_means = []
spectral_learned_stds = []

for k in K_values:
    if k == 1:
        # Silhouette score is undefined for a single cluster
        spectral_learned_means.append(np.nan)
        spectral_learned_stds.append(np.nan)
        print("K=1: Silhouette score NaN it requires at least 2 clusters")
    else:
        scores = []
        for seed in range(10):
            # Run Spectral Clustering with default parameters and a different random seed each time
            sc = SpectralClustering(n_clusters=k, random_state=seed)
            labels = sc.fit_predict(learned_np_sub)

            # Compute Silhouette Coefficient for this run
            score = silhouette_score(learned_np_sub, labels)
            scores.append(score)
            
            # want to see the iteration if it running
            print(f"K={k}, Seed={seed+1}/10 done")

        # Store mean and std across 10 runs
        spectral_learned_means.append(np.mean(scores))
        spectral_learned_stds.append(np.std(scores))
        print(f"K={k}: Mean={np.mean(scores):.4f} | Std={np.std(scores):.4f}")

print("Spectral Clustering on Learned Representations complete!")
```
